## 2. Extract Project Files
Upload the `spacetech_project.zip` file to your Kaggle input datasets, or directly upload it to the `/kaggle/working/` directory. Then run the block below to extract it.

```python
import os
import zipfile

# Ensure you uploaded the zip file. Update this path if you uploaded it as a dataset.
zip_path = "/kaggle/working/spacetech_project.zip"
extract_dir = "/kaggle/working/spacetech"

if os.path.exists(zip_path):
    print(f"Extracting {zip_path}...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)
    print("Extraction complete.")
else:
    print(f"Zip file not found at {zip_path}. Please upload it!")
```

## 4. Save Checkpoints
When training finishes or is interrupted, make sure to save the `checkpoints/` and `.zip` files so they persist in Kaggle's output directory.

```python
import shutil

# Copy checkpoints back to working root to ensure Kaggle saves them
if os.path.exists("checkpoints"):
    shutil.copytree("checkpoints", "/kaggle/working/checkpoints_backup", dirs_exist_ok=True)
    print("Saved checkpoints.")

for file in os.listdir("."):
    if file.endswith(".zip") and file != "spacetech_project.zip":
        shutil.copy(file, "/kaggle/working/")
        print(f"Saved model {file}.")
```
